```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array([2026, 2027, 2028])

# 多因素收入模型（单位：万元）
base_revenue = np.array([8, 10, 12])        # 基础销售
design_orders = np.array([3, 4, 6])         # 定制订单
ip_licensing = np.array([2, 3, 5])          # IP授权
edu_projects = np.array([2, 1, 2])          # 研学/合作

revenue = base_revenue + design_orders + ip_licensing + edu_projects

profit = revenue * np.array([0.72, 0.68, 0.64])  # 利润率略下降（规模扩张成本）

plt.figure(figsize=(8,5))
plt.plot(years, revenue, marker='o', label="Total Revenue")
plt.plot(years, profit, marker='o', label="Profit")

plt.title("Multi-factor Financial Forecast (Revenue & Profit)")
plt.xlabel("Year")
plt.ylabel("万元")
plt.legend()
plt.grid()
plt.show()
```

```python
years = np.array([2026, 2027, 2028])

material_cost = np.array([1.0, 0.85, 0.72])
labor_cost = np.array([1.0, 0.90, 0.80])
energy_cost = np.array([1.0, 0.88, 0.75])

total_cost_index = (material_cost + labor_cost + energy_cost) / 3

plt.figure(figsize=(8,5))
plt.plot(years, material_cost, marker='o', label="Material Cost")
plt.plot(years, labor_cost, marker='o', label="Labor Cost")
plt.plot(years, energy_cost, marker='o', label="Energy Cost")
plt.plot(years, total_cost_index, marker='o', linewidth=3, label="Total Cost Index")

plt.title("Cost Optimization Model (Multi-factor)")
plt.xlabel("Year")
plt.ylabel("Relative Cost Index")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

years = np.array([2026, 2027, 2028])

base_margin = np.array([0.15, 0.18, 0.20])
ip_effect = np.array([0.00, 0.03, 0.06])
cost_effect = np.array([0.00, 0.02, 0.04])
premium_product_effect = np.array([0.00, 0.02, 0.05])

plt.figure(figsize=(8,5))

plt.bar(years, base_margin, label="Base Margin")
plt.bar(years, ip_effect, bottom=base_margin, label="IP Effect")
plt.bar(years, cost_effect, bottom=base_margin+ip_effect, label="Cost Optimization")
plt.bar(years, premium_product_effect, 
        bottom=base_margin+ip_effect+cost_effect, 
        label="Product Premium Effect")

plt.title("Profit Margin Composition Analysis")
plt.xlabel("Year")
plt.ylabel("Profit Margin")
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.show()
```
