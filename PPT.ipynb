```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager
from pathlib import Path


# =====================================================
# 1. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"
OUTPUT_PATH = "multi_parameter_process_optimization_curve.png"


# =====================================================
# 2. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    for font in candidate_fonts:
        if font in available_fonts:
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False


set_chinese_font()


# =====================================================
# 3. 读取数据
# =====================================================

df = pd.read_csv(CSV_PATH)

numeric_cols = [
    "record_id",
    "experiment_round",
    "days_since_first",
    "temperature_c",
    "maker_count",
    "overheat_score",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overall_quality_score_100",
    "hold_time_min"
]

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()


# =====================================================
# 4. 指标归一化
# 说明：
# particle_score / volume_score / transparency_score / success_score 是 0-5 分制
# overall_quality_score_100 是 0-100 分制
# overheat_score 也是 0-5 风险分制，这里转换成 0-100 的风险值
# =====================================================

df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_risk_norm"] = df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]


# =====================================================
# 5. 按温度汇总
# =====================================================

summary = (
    df.groupby("temperature_c")
    .agg(
        sample_count=("record_id", "count"),
        maker_count_mean=("maker_count", "mean"),
        particle_mean=("particle_norm", "mean"),
        volume_mean=("volume_norm", "mean"),
        transparency_mean=("transparency_norm", "mean"),
        success_mean=("success_norm", "mean"),
        overheat_risk_mean=("overheat_risk_norm", "mean"),
        overall_mean=("overall_norm", "mean"),
        overall_std=("overall_norm", "std"),
        overall_max=("overall_norm", "max"),
        overall_min=("overall_norm", "min"),
        round_min=("experiment_round", "min"),
        round_max=("experiment_round", "max")
    )
    .reset_index()
    .sort_values("temperature_c")
)

summary["overall_std"] = summary["overall_std"].fillna(0)

x = summary["temperature_c"].values


# =====================================================
# 6. 拟合函数
# 数据点较少，采用二次拟合形成平滑趋势
# =====================================================

def smooth_curve(x, y, points=200):
    x = np.asarray(x)
    y = np.asarray(y)

    if len(x) >= 3:
        degree = 2
    else:
        degree = 1

    coef = np.polyfit(x, y, degree)
    poly = np.poly1d(coef)

    x_fit = np.linspace(x.min(), x.max(), points)
    y_fit = poly(x_fit)

    return x_fit, y_fit


# =====================================================
# 7. 绘制单张多参数复杂图
# =====================================================

fig, ax = plt.subplots(figsize=(15.5, 8.5), dpi=300)

# 背景区间
ax.axvspan(755, 765, alpha=0.10, label="推荐低温区间 / Recommended Low-Temp Zone")
ax.axvspan(775, 785, alpha=0.07, label="过渡区间 / Transition Zone")
ax.axvspan(795, 805, alpha=0.07, label="过热风险区间 / Overheating Risk Zone")

# 多指标线条配置
series_config = [
    ("overall_mean", "综合质量 Overall Quality", 4.2, "-"),
    ("particle_mean", "颗粒感 Particle Texture", 2.4, "-"),
    ("volume_mean", "体积感 Volume Form", 2.4, "-"),
    ("transparency_mean", "透光度 Transparency", 2.4, "-"),
    ("success_mean", "成功度 Success", 2.4, "-"),
    ("overheat_risk_mean", "过热风险 Overheating Risk", 2.8, "--")
]

# 绘制多参数曲线
for col, label, linewidth, linestyle in series_config:
    y = summary[col].values
    x_fit, y_fit = smooth_curve(x, y)

    ax.plot(
        x_fit,
        y_fit,
        linewidth=linewidth,
        linestyle=linestyle,
        label=label
    )

    ax.scatter(
        x,
        y,
        s=90,
        edgecolors="black",
        linewidths=0.8,
        zorder=5
    )

# 综合质量误差带
ax.fill_between(
    x,
    summary["overall_mean"] - summary["overall_std"],
    summary["overall_mean"] + summary["overall_std"],
    alpha=0.15,
    label="综合质量波动 / Quality Std. Range"
)

# 样本数量气泡，叠加在综合质量点附近
bubble_size = summary["sample_count"] * 55

ax.scatter(
    x,
    summary["overall_mean"],
    s=bubble_size,
    facecolors="none",
    edgecolors="black",
    linewidths=2,
    zorder=6,
    label="样本数量 / Sample Count"
)

# 标注每个温度的综合质量和样本数
for _, row in summary.iterrows():
    ax.text(
        row["temperature_c"],
        row["overall_mean"] + 5,
        f'{int(row["temperature_c"])}℃\n质量 {row["overall_mean"]:.1f}\n样本 n={int(row["sample_count"])}',
        ha="center",
        va="bottom",
        fontsize=10.5,
        fontweight="bold",
        bbox=dict(
            boxstyle="round,pad=0.35",
            facecolor="white",
            edgecolor="gray",
            alpha=0.88
        )
    )

# 最优温度标注
best_row = summary.loc[summary["overall_mean"].idxmax()]
best_temp = best_row["temperature_c"]
best_quality = best_row["overall_mean"]

ax.annotate(
    f"最优温度 / Best Temperature\n{int(best_temp)}℃ 综合质量 {best_quality:.1f}",
    xy=(best_temp, best_quality),
    xytext=(best_temp + 10, best_quality - 18),
    fontsize=12,
    fontweight="bold",
    arrowprops=dict(
        arrowstyle="->",
        linewidth=2
    ),
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="black",
        alpha=0.92
    )
)

# 过热风险标注
hot_row = summary.loc[summary["temperature_c"].idxmax()]
hot_temp = hot_row["temperature_c"]

ax.annotate(
    "高温区：过热风险上升\nHigh Temp: Overheating Risk",
    xy=(hot_temp, hot_row["overheat_risk_mean"]),
    xytext=(hot_temp - 22, 72),
    fontsize=11,
    fontweight="bold",
    arrowprops=dict(
        arrowstyle="->",
        linewidth=1.8
    ),
    bbox=dict(
        boxstyle="round,pad=0.4",
        facecolor="white",
        edgecolor="gray",
        alpha=0.90
    )
)

# 标题和轴
ax.set_title(
    "降耗智造：再生玻璃低温烧制多参数工艺优化曲线\n"
    "Energy-Saving Smart Manufacturing: Multi-Parameter Optimization Curve for Recycled Glass Fusing",
    fontsize=20,
    fontweight="bold",
    pad=24
)

ax.set_xlabel(
    "烧制温度 / Fusing Temperature (℃)",
    fontsize=13,
    fontweight="bold"
)

ax.set_ylabel(
    "归一化指标得分 / Normalized Indicator Score (0-100)",
    fontsize=13,
    fontweight="bold"
)

ax.set_xticks(sorted(summary["temperature_c"].unique()))
ax.set_ylim(0, 110)
ax.grid(True, linestyle="--", alpha=0.35)

# 图例
ax.legend(
    loc="upper center",
    bbox_to_anchor=(0.5, -0.12),
    ncol=3,
    fontsize=10.5,
    frameon=True
)

# 结论文本框
conclusion = (
    "核心结论 / Key Finding:\n"
    "760℃ shows the best overall performance, with particle texture, volume form, transparency and success score rising together.\n"
    "760℃下综合质量最优，多项成品指标同步提升；800℃存在过热风险，质量表现明显下降。"
)

ax.text(
    0.02,
    0.04,
    conclusion,
    transform=ax.transAxes,
    fontsize=10.8,
    va="bottom",
    ha="left",
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="white",
        edgecolor="gray",
        alpha=0.92
    )
)

# 页脚
fig.text(
    0.5,
    0.015,
    "Data source: glass_experiment_numeric_only.csv | 数据来源：再生玻璃热熔实验记录",
    ha="center",
    fontsize=10.5
)

plt.tight_layout(rect=[0, 0.07, 1, 0.95])
plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight")
plt.show()

print(f"已生成单张多参数复杂图表：{OUTPUT_PATH}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager


# =====================================================
# 1. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"
OUTPUT_PATH = "glass_fusing_heatmap_optimized.png"


# =====================================================
# 2. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    for font in candidate_fonts:
        if font in available_fonts:
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False


set_chinese_font()


# =====================================================
# 3. 读取数据
# =====================================================

df = pd.read_csv(CSV_PATH)

required_cols = [
    "temperature_c",
    "overheat_score",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overall_quality_score_100"
]

for col in required_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()


# =====================================================
# 4. 指标处理
# =====================================================
# 说明：
# particle_score、volume_score、transparency_score、success_score、overheat_score 是 0-5 分制
# overall_quality_score_100 是 0-100 分制
# 为了热力图可比较，将前五项统一转成 0-100
# 过热风险是负向指标，数值越高越不好，所以额外生成“低过热表现分”
# 即：overheat_control_score = 100 - overheat_score / 5 * 100
# 这样热力图中所有指标都变成“越高越好”

df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_control_norm"] = 100 - df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]


# =====================================================
# 5. 按温度汇总
# =====================================================

metric_cols = [
    "particle_norm",
    "volume_norm",
    "transparency_norm",
    "success_norm",
    "overheat_control_norm",
    "overall_norm"
]

raw_cols = [
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overheat_score",
    "overall_quality_score_100"
]

metric_labels = [
    "颗粒感\nParticle",
    "体积感\nVolume",
    "透光度\nTransparency",
    "成功度\nSuccess",
    "控过热\nOverheat Control",
    "综合质量\nOverall Quality"
]

temp_order = sorted(df["temperature_c"].unique())

heatmap_norm = (
    df.groupby("temperature_c")[metric_cols]
    .mean()
    .loc[temp_order]
)

heatmap_raw = (
    df.groupby("temperature_c")[raw_cols]
    .mean()
    .loc[temp_order]
)

sample_count = (
    df.groupby("temperature_c")["temperature_c"]
    .count()
    .loc[temp_order]
)


# =====================================================
# 6. 绘制热力图
# =====================================================

fig, ax = plt.subplots(figsize=(13.5, 7.2), dpi=300)

im = ax.imshow(
    heatmap_norm.values,
    aspect="auto",
    vmin=0,
    vmax=100,
    cmap="RdYlGn"
)

# 标题
ax.set_title(
    "再生玻璃烧制多指标性能热力图\n"
    "Multi-Metric Performance Heatmap of Recycled Glass Fusing",
    fontsize=20,
    fontweight="bold",
    pad=22
)

# 坐标轴
ax.set_xticks(np.arange(len(metric_labels)))
ax.set_xticklabels(metric_labels, fontsize=11)

ax.set_yticks(np.arange(len(temp_order)))
ax.set_yticklabels(
    [f"{int(t)}℃\n样本 n={int(sample_count.loc[t])}" for t in temp_order],
    fontsize=12
)

ax.set_xlabel(
    "性能指标 / Performance Metrics",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

ax.set_ylabel(
    "烧制温度 / Fusing Temperature",
    fontsize=13,
    fontweight="bold",
    labelpad=12
)

# 格子线
ax.set_xticks(np.arange(-0.5, len(metric_cols), 1), minor=True)
ax.set_yticks(np.arange(-0.5, len(temp_order), 1), minor=True)

ax.grid(
    which="minor",
    color="white",
    linestyle="-",
    linewidth=2
)

ax.tick_params(which="minor", bottom=False, left=False)


# =====================================================
# 7. 单元格数字标注
# =====================================================

for i, temp in enumerate(temp_order):
    for j, col in enumerate(metric_cols):
        norm_value = heatmap_norm.loc[temp, col]

        # 原始分值显示
        if col == "overall_norm":
            raw_text = f"{heatmap_raw.loc[temp, 'overall_quality_score_100']:.1f}"
        elif col == "overheat_control_norm":
            # 控过热指标显示方式：控过热得分，同时括号里写原过热分
            raw_overheat = heatmap_raw.loc[temp, "overheat_score"]
            raw_text = f"{norm_value:.0f}\n过热{raw_overheat:.1f}/5"
        else:
            raw_col = raw_cols[j]
            raw_text = f"{heatmap_raw.loc[temp, raw_col]:.1f}/5"

        text_color = "white" if norm_value < 45 else "black"

        ax.text(
            j,
            i,
            raw_text,
            ha="center",
            va="center",
            fontsize=10.5,
            fontweight="bold",
            color=text_color
        )


# =====================================================
# 8. 标出最优温度
# =====================================================

best_temp = heatmap_norm["overall_norm"].idxmax()
best_row_idx = temp_order.index(best_temp)

ax.add_patch(
    plt.Rectangle(
        (-0.5, best_row_idx - 0.5),
        len(metric_cols),
        1,
        fill=False,
        edgecolor="black",
        linewidth=3.0
    )
)

ax.text(
    len(metric_cols) - 0.1,
    best_row_idx,
    "  最优区间\n  Best Zone",
    ha="left",
    va="center",
    fontsize=11,
    fontweight="bold",
    bbox=dict(
        boxstyle="round,pad=0.35",
        facecolor="white",
        edgecolor="black",
        alpha=0.92
    )
)


# =====================================================
# 9. 色条说明
# =====================================================

cbar = fig.colorbar(im, ax=ax, fraction=0.035, pad=0.035)

cbar.set_label(
    "归一化表现 / Normalized Performance\n绿色越深表示综合表现越优",
    fontsize=11,
    labelpad=12
)

cbar.set_ticks([0, 25, 50, 75, 100])
cbar.set_ticklabels([
    "低\nLow",
    "偏低",
    "中等\nMedium",
    "较优",
    "高\nHigh"
])


# =====================================================
# 10. 图下注释
# =====================================================

note = (
    "注：颗粒感、体积感、透光度、成功度与控过热指标统一换算为0-100分；"
    "控过热为反向处理指标，数值越高表示过热风险越低。\n"
    "Note: Metrics are normalized to 0-100. Overheat control is reversed, where higher values mean lower overheating risk."
)

fig.text(
    0.5,
    0.02,
    note,
    ha="center",
    fontsize=10
)

plt.tight_layout(rect=[0, 0.08, 1, 0.95])
plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight")
plt.show()

print(f"热力图已保存：{OUTPUT_PATH}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager


# =====================================================
# 1. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"
OUTPUT_PATH = "production_process_decision_radar.png"


# =====================================================
# 2. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    for font in candidate_fonts:
        if font in available_fonts:
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False


set_chinese_font()


# =====================================================
# 3. 读取数据
# =====================================================

df = pd.read_csv(CSV_PATH)

required_cols = [
    "temperature_c",
    "overheat_score",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overall_quality_score_100",
    "record_id"
]

for col in required_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()


# =====================================================
# 4. 指标归一化
# =====================================================
# 说明：
# 颗粒感、体积感、透光度、成功度、过热分为 0-5 分制
# 综合质量为 0-100 分制
# 过热是负向指标，所以转为“控过热能力”
# 控过热能力 = 100 - 过热分 / 5 * 100
# 这样所有指标都变成：越高越好

df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_control_norm"] = 100 - df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]


# =====================================================
# 5. 按温度汇总
# =====================================================

metric_cols = [
    "particle_norm",
    "volume_norm",
    "transparency_norm",
    "success_norm",
    "overheat_control_norm",
    "overall_norm"
]

metric_labels = [
    "颗粒感\nParticle",
    "体积感\nVolume",
    "透光度\nTransparency",
    "成功度\nSuccess",
    "控过热\nOverheat Control",
    "综合质量\nOverall"
]

summary = (
    df.groupby("temperature_c")
    .agg(
        sample_count=("record_id", "count"),
        particle_norm=("particle_norm", "mean"),
        volume_norm=("volume_norm", "mean"),
        transparency_norm=("transparency_norm", "mean"),
        success_norm=("success_norm", "mean"),
        overheat_control_norm=("overheat_control_norm", "mean"),
        overall_norm=("overall_norm", "mean")
    )
    .reset_index()
    .sort_values("temperature_c")
)

temp_order = sorted(summary["temperature_c"].unique())


# =====================================================
# 6. 构造雷达图数据
# =====================================================

num_metrics = len(metric_cols)

angles = np.linspace(0, 2 * np.pi, num_metrics, endpoint=False).tolist()
angles += angles[:1]


# =====================================================
# 7. 绘制雷达图
# =====================================================

fig = plt.figure(figsize=(11.5, 9), dpi=300)
ax = plt.subplot(111, polar=True)

# 背景网格
ax.set_theta_offset(np.pi / 2)
ax.set_theta_direction(-1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(metric_labels, fontsize=12, fontweight="bold")

ax.set_ylim(0, 100)
ax.set_yticks([20, 40, 60, 80, 100])
ax.set_yticklabels(["20", "40", "60", "80", "100"], fontsize=9)
ax.grid(True, linestyle="--", alpha=0.45)


# =====================================================
# 8. 绘制每个温度的多指标轮廓
# =====================================================

for temp in temp_order:
    row = summary[summary["temperature_c"] == temp].iloc[0]

    values = [row[col] for col in metric_cols]
    values += values[:1]

    label = f"{int(temp)}℃  n={int(row['sample_count'])}"

    ax.plot(
        angles,
        values,
        linewidth=2.8,
        marker="o",
        markersize=5,
        label=label
    )

    ax.fill(
        angles,
        values,
        alpha=0.10
    )


# =====================================================
# 9. 标注最优温度
# =====================================================

best_row = summary.loc[summary["overall_norm"].idxmax()]
best_temp = int(best_row["temperature_c"])
best_score = best_row["overall_norm"]

ax.text(
    0,
    -16,
    f"推荐工艺温度 / Recommended Process Temperature：{best_temp}℃\n"
    f"综合质量均值 / Mean Overall Quality：{best_score:.1f}",
    ha="center",
    va="center",
    fontsize=13,
    fontweight="bold",
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="white",
        edgecolor="black",
        alpha=0.92
    )
)


# =====================================================
# 10. 标题与说明
# =====================================================

plt.title(
    "生产端智能工艺决策雷达图\n"
    "Production-Side Intelligent Process Decision Radar",
    fontsize=20,
    fontweight="bold",
    pad=38
)

ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.25, 1.12),
    fontsize=11,
    frameon=True
)


# =====================================================
# 11. 图下注释
# =====================================================

note = (
    "说明：所有指标统一换算为0-100分；控过热为反向处理指标，数值越高表示过热风险越低。\n"
    "Note: All metrics are normalized to 0-100. Overheat control is reversed, where higher values indicate lower overheating risk."
)

fig.text(
    0.5,
    0.035,
    note,
    ha="center",
    fontsize=10.5
)

plt.tight_layout(rect=[0, 0.08, 1, 0.95])
plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight")
plt.show()

print(f"图表已保存：{OUTPUT_PATH}")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter, FFMpegWriter
from matplotlib import font_manager
from PIL import Image


# =====================================================
# 1. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"

OUTPUT_MP4 = "animated_glass_process_curve.mp4"
OUTPUT_GIF = "animated_glass_process_curve.gif"

# 是否同时生成 GIF
# 只想要 MP4 就保持 False
# 同时想要 GIF 就改成 True
GENERATE_GIF = False


# =====================================================
# 2. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    selected_font = None

    for font in candidate_fonts:
        if font in available_fonts:
            selected_font = font
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False

    if selected_font:
        print(f"当前使用字体：{selected_font}")
    else:
        print("未找到常见中文字体，若中文乱码，请安装 Microsoft YaHei 或 SimHei。")


set_chinese_font()


# =====================================================
# 3. 检查 CSV 文件
# =====================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"没有找到数据文件：{CSV_PATH}\n"
        f"请把 glass_experiment_numeric_only.csv 放到当前 Python 脚本同一目录。"
    )


# =====================================================
# 4. 读取与处理数据
# =====================================================

df = pd.read_csv(CSV_PATH)

required_cols = [
    "temperature_c",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overheat_score",
    "overall_quality_score_100"
]

missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(
        "CSV 缺少以下必要字段：\n"
        + "\n".join(missing_cols)
    )

for col in required_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()

if df.empty:
    raise ValueError("数据为空，请检查 CSV 中 temperature_c 和 overall_quality_score_100 是否有有效数值。")


# =====================================================
# 5. 指标归一化
# =====================================================
# particle_score / volume_score / transparency_score / success_score / overheat_score
# 原始是 0-5 分制
# overall_quality_score_100 原始是 0-100 分制
#
# 过热是负向指标，所以转换为“控过热能力”
# 分数越高，说明过热风险越低
# =====================================================

df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_control_norm"] = 100 - df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]


# =====================================================
# 6. 按温度汇总
# =====================================================

summary = (
    df.groupby("temperature_c")
    .agg(
        sample_count=("temperature_c", "count"),
        particle=("particle_norm", "mean"),
        volume=("volume_norm", "mean"),
        transparency=("transparency_norm", "mean"),
        success=("success_norm", "mean"),
        overheat_control=("overheat_control_norm", "mean"),
        overall=("overall_norm", "mean")
    )
    .reset_index()
)

# 关键：按 800 → 780 → 760 的顺序展示
summary = summary.sort_values("temperature_c", ascending=False).reset_index(drop=True)

x = summary["temperature_c"].values

if len(x) < 2:
    raise ValueError("温度种类少于 2 个，无法生成动态折线。")


series = {
    "综合质量 Overall": summary["overall"].values,
    "颗粒感 Particle": summary["particle"].values,
    "体积感 Volume": summary["volume"].values,
    "透光度 Transparency": summary["transparency"].values,
    "成功度 Success": summary["success"].values,
    "控过热 Overheat Control": summary["overheat_control"].values,
}


# =====================================================
# 7. 创建动画画布
# =====================================================

fig, ax = plt.subplots(figsize=(12, 6.8), dpi=160)

# x 轴从高温到低温展示
ax.set_xlim(x.max() + 5, x.min() - 5)
ax.set_ylim(0, 110)

ax.set_xlabel(
    "烧制温度 / Fusing Temperature (℃)",
    fontsize=12,
    fontweight="bold"
)

ax.set_ylabel(
    "归一化指标得分 / Normalized Score (0-100)",
    fontsize=12,
    fontweight="bold"
)

ax.set_title(
    "低温烧制多参数动态优化曲线\nAnimated Multi-Parameter Optimization Curve",
    fontsize=18,
    fontweight="bold",
    pad=18
)

ax.grid(True, linestyle="--", alpha=0.35)

# 标出三个温度点
ax.set_xticks(x)
ax.set_xticklabels([f"{int(t)}℃" for t in x], fontsize=11)

lines = {}
points = {}

for name in series:
    line, = ax.plot([], [], linewidth=2.8, label=name)
    point, = ax.plot([], [], "o", markersize=7)
    lines[name] = line
    points[name] = point

ax.legend(loc="upper right", fontsize=9, frameon=True)

note = ax.text(
    0.02,
    0.04,
    "",
    transform=ax.transAxes,
    fontsize=11,
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="gray",
        alpha=0.9
    )
)


# =====================================================
# 8. 动画逻辑
# =====================================================

frames_per_segment = 24
hold_frames = 36
total_frames = (len(x) - 1) * frames_per_segment + hold_frames


def interpolate_values(y, frame):
    """
    让折线从 800℃ → 780℃ → 760℃ 逐段出现。
    """

    max_dynamic_frame = (len(x) - 1) * frames_per_segment

    if frame >= max_dynamic_frame:
        return x, y

    segment = frame // frames_per_segment
    progress = (frame % frames_per_segment) / frames_per_segment

    x_current = list(x[:segment + 1])
    y_current = list(y[:segment + 1])

    x_next = x[segment] + (x[segment + 1] - x[segment]) * progress
    y_next = y[segment] + (y[segment + 1] - y[segment]) * progress

    x_current.append(x_next)
    y_current.append(y_next)

    return np.array(x_current), np.array(y_current)


def get_note_text(frame):
    if frame < frames_per_segment:
        return "800℃：高温区过热风险明显，多项指标偏低"

    if frame < frames_per_segment * 2:
        return "780℃：较800℃略有改善，但整体稳定性仍不足"

    return "760℃：综合质量与多项指标同步提升，形成较优低温烧制区间"


def update(frame):
    artists = []

    for name, y in series.items():
        x_current, y_current = interpolate_values(y, frame)

        lines[name].set_data(x_current, y_current)
        points[name].set_data(x_current, y_current)

        artists.extend([lines[name], points[name]])

    note.set_text("动态结论 / Dynamic Finding：\n" + get_note_text(frame))
    artists.append(note)

    return artists


ani = FuncAnimation(
    fig,
    update,
    frames=total_frames,
    interval=90,
    blit=True,
    repeat=True
)


# =====================================================
# 9. 生成 MP4
# =====================================================

print("正在生成 MP4，请稍等...")

try:
    import imageio_ffmpeg

    ffmpeg_path = imageio_ffmpeg.get_ffmpeg_exe()
    plt.rcParams["animation.ffmpeg_path"] = ffmpeg_path

    mp4_writer = FFMpegWriter(
        fps=12,
        bitrate=2200,
        codec="libx264",
        extra_args=["-pix_fmt", "yuv420p"]
    )

    ani.save(OUTPUT_MP4, writer=mp4_writer)

    print(f"MP4 已生成：{OUTPUT_MP4}")
    print(f"使用的 ffmpeg 路径：{ffmpeg_path}")

except Exception as e:
    print("MP4 生成失败。")
    print("原因：", e)
    print("解决办法：请先运行 pip install imageio-ffmpeg")
    print("如果仍失败，再生成 GIF 作为备用。")


# =====================================================
# 10. 可选生成 GIF
# =====================================================

if GENERATE_GIF:
    print("\n正在生成 GIF，请稍等...")

    try:
        gif_writer = PillowWriter(fps=12)
        ani.save(OUTPUT_GIF, writer=gif_writer)

        print(f"GIF 已生成：{OUTPUT_GIF}")

    except Exception as e:
        print("GIF 生成失败。")
        print("原因：", e)


plt.close(fig)


# =====================================================
# 11. 检查 MP4 是否生成成功
# =====================================================

print("\n正在检查输出文件...")

if os.path.exists(OUTPUT_MP4):
    mp4_size_mb = os.path.getsize(OUTPUT_MP4) / 1024 / 1024

    if mp4_size_mb > 0:
        print(f"MP4 检查通过：{OUTPUT_MP4}")
        print(f"MP4 文件大小：{mp4_size_mb:.2f} MB")
    else:
        print("MP4 文件存在，但大小为 0，生成失败。")
else:
    print("没有检测到 MP4 文件。")


# =====================================================
# 12. 检查 GIF 是否是真动图
# =====================================================

if GENERATE_GIF and os.path.exists(OUTPUT_GIF):
    try:
        img = Image.open(OUTPUT_GIF)

        frame_count = getattr(img, "n_frames", 1)
        file_format = img.format
        file_size_mb = os.path.getsize(OUTPUT_GIF) / 1024 / 1024

        print("\nGIF 文件格式：", file_format)
        print("GIF 帧数：", frame_count)
        print(f"GIF 文件大小：{file_size_mb:.2f} MB")

        if file_format == "GIF" and frame_count > 1:
            print("GIF 检查通过：这是可以播放的 GIF 动图。")
        else:
            print("GIF 检查失败：这不是有效 GIF 动图。")

    except Exception as e:
        print("GIF 检查失败。")
        print("原因：", e)


# =====================================================
# 13. 输出数据摘要
# =====================================================

print("\n温度汇总数据：")
print(summary.round(2).to_string(index=False))


# =====================================================
# 14. 使用建议
# =====================================================

print("\n使用建议：")
print("1. 发给学姐时，优先发送 animated_glass_process_curve.mp4。")
print("2. PPT 中建议插入 MP4，播放更稳定。")
print("3. 如果必须使用 GIF，把 GENERATE_GIF 改成 True 后重新运行。")
print("4. 不要把静态截图改后缀成 .gif，那样不会动。")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter, FFMpegWriter
from matplotlib import font_manager
from PIL import Image


# =====================================================
# 1. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"

OUTPUT_MP4 = "animated_glass_process_curve.mp4"
OUTPUT_GIF = "animated_glass_process_curve.gif"

# 是否同时生成 GIF
# 只想要 MP4 就保持 False
# 同时想要 GIF 就改成 True
GENERATE_GIF = False


# =====================================================
# 2. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    selected_font = None

    for font in candidate_fonts:
        if font in available_fonts:
            selected_font = font
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False

    if selected_font:
        print(f"当前使用字体：{selected_font}")
    else:
        print("未找到常见中文字体，若中文乱码，请安装 Microsoft YaHei 或 SimHei。")


set_chinese_font()


# =====================================================
# 3. 检查 CSV 文件
# =====================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"没有找到数据文件：{CSV_PATH}\n"
        f"请把 glass_experiment_numeric_only.csv 放到当前 Python 脚本同一目录。"
    )


# =====================================================
# 4. 读取与处理数据
# =====================================================

df = pd.read_csv(CSV_PATH)

required_cols = [
    "temperature_c",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overheat_score",
    "overall_quality_score_100"
]

missing_cols = [col for col in required_cols if col not in df.columns]

if missing_cols:
    raise ValueError(
        "CSV 缺少以下必要字段：\n"
        + "\n".join(missing_cols)
    )

for col in required_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()

if df.empty:
    raise ValueError("数据为空，请检查 CSV 中 temperature_c 和 overall_quality_score_100 是否有有效数值。")


# =====================================================
# 5. 指标归一化
# =====================================================
# particle_score / volume_score / transparency_score / success_score / overheat_score
# 原始是 0-5 分制
# overall_quality_score_100 原始是 0-100 分制
#
# 过热是负向指标，所以转换为“控过热能力”
# 分数越高，说明过热风险越低
# =====================================================

df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_control_norm"] = 100 - df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]


# =====================================================
# 6. 按温度汇总
# =====================================================

summary = (
    df.groupby("temperature_c")
    .agg(
        sample_count=("temperature_c", "count"),
        particle=("particle_norm", "mean"),
        volume=("volume_norm", "mean"),
        transparency=("transparency_norm", "mean"),
        success=("success_norm", "mean"),
        overheat_control=("overheat_control_norm", "mean"),
        overall=("overall_norm", "mean")
    )
    .reset_index()
)

# 关键：按 800 → 780 → 760 的顺序展示
summary = summary.sort_values("temperature_c", ascending=False).reset_index(drop=True)

x = summary["temperature_c"].values

if len(x) < 2:
    raise ValueError("温度种类少于 2 个，无法生成动态折线。")


series = {
    "综合质量 Overall": summary["overall"].values,
    "颗粒感 Particle": summary["particle"].values,
    "体积感 Volume": summary["volume"].values,
    "透光度 Transparency": summary["transparency"].values,
    "成功度 Success": summary["success"].values,
    "控过热 Overheat Control": summary["overheat_control"].values,
}


# =====================================================
# 7. 创建动画画布
# =====================================================

fig, ax = plt.subplots(figsize=(12, 6.8), dpi=160)

# x 轴从高温到低温展示
ax.set_xlim(x.max() + 5, x.min() - 5)
ax.set_ylim(0, 110)

ax.set_xlabel(
    "烧制温度 / Fusing Temperature (℃)",
    fontsize=12,
    fontweight="bold"
)

ax.set_ylabel(
    "归一化指标得分 / Normalized Score (0-100)",
    fontsize=12,
    fontweight="bold"
)

ax.set_title(
    "低温烧制多参数动态优化曲线\nAnimated Multi-Parameter Optimization Curve",
    fontsize=18,
    fontweight="bold",
    pad=18
)

ax.grid(True, linestyle="--", alpha=0.35)

# 标出三个温度点
ax.set_xticks(x)
ax.set_xticklabels([f"{int(t)}℃" for t in x], fontsize=11)

lines = {}
points = {}

for name in series:
    line, = ax.plot([], [], linewidth=2.8, label=name)
    point, = ax.plot([], [], "o", markersize=7)
    lines[name] = line
    points[name] = point

ax.legend(loc="upper right", fontsize=9, frameon=True)

note = ax.text(
    0.02,
    0.04,
    "",
    transform=ax.transAxes,
    fontsize=11,
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="gray",
        alpha=0.9
    )
)


# =====================================================
# 8. 动画逻辑
# =====================================================

frames_per_segment = 24
hold_frames = 36
total_frames = (len(x) - 1) * frames_per_segment + hold_frames


def interpolate_values(y, frame):
    """
    让折线从 800℃ → 780℃ → 760℃ 逐段出现。
    """

    max_dynamic_frame = (len(x) - 1) * frames_per_segment

    if frame >= max_dynamic_frame:
        return x, y

    segment = frame // frames_per_segment
    progress = (frame % frames_per_segment) / frames_per_segment

    x_current = list(x[:segment + 1])
    y_current = list(y[:segment + 1])

    x_next = x[segment] + (x[segment + 1] - x[segment]) * progress
    y_next = y[segment] + (y[segment + 1] - y[segment]) * progress

    x_current.append(x_next)
    y_current.append(y_next)

    return np.array(x_current), np.array(y_current)


def get_note_text(frame):
    if frame < frames_per_segment:
        return "800℃：高温区过热风险明显，多项指标偏低"

    if frame < frames_per_segment * 2:
        return "780℃：较800℃略有改善，但整体稳定性仍不足"

    return "760℃：综合质量与多项指标同步提升，形成较优低温烧制区间"


def update(frame):
    artists = []

    for name, y in series.items():
        x_current, y_current = interpolate_values(y, frame)

        lines[name].set_data(x_current, y_current)
        points[name].set_data(x_current, y_current)

        artists.extend([lines[name], points[name]])

    note.set_text("动态结论 / Dynamic Finding：\n" + get_note_text(frame))
    artists.append(note)

    return artists


ani = FuncAnimation(
    fig,
    update,
    frames=total_frames,
    interval=90,
    blit=True,
    repeat=True
)


# =====================================================
# 9. 生成 MP4
# =====================================================

print("正在生成 MP4，请稍等...")

try:
    import imageio_ffmpeg

    ffmpeg_path = imageio_ffmpeg.get_ffmpeg_exe()
    plt.rcParams["animation.ffmpeg_path"] = ffmpeg_path

    mp4_writer = FFMpegWriter(
        fps=12,
        bitrate=2200,
        codec="libx264",
        extra_args=["-pix_fmt", "yuv420p"]
    )

    ani.save(OUTPUT_MP4, writer=mp4_writer)

    print(f"MP4 已生成：{OUTPUT_MP4}")
    print(f"使用的 ffmpeg 路径：{ffmpeg_path}")

except Exception as e:
    print("MP4 生成失败。")
    print("原因：", e)
    print("解决办法：请先运行 pip install imageio-ffmpeg")
    print("如果仍失败，再生成 GIF 作为备用。")


# =====================================================
# 10. 可选生成 GIF
# =====================================================

if GENERATE_GIF:
    print("\n正在生成 GIF，请稍等...")

    try:
        gif_writer = PillowWriter(fps=12)
        ani.save(OUTPUT_GIF, writer=gif_writer)

        print(f"GIF 已生成：{OUTPUT_GIF}")

    except Exception as e:
        print("GIF 生成失败。")
        print("原因：", e)


plt.close(fig)


# =====================================================
# 11. 检查 MP4 是否生成成功
# =====================================================

print("\n正在检查输出文件...")

if os.path.exists(OUTPUT_MP4):
    mp4_size_mb = os.path.getsize(OUTPUT_MP4) / 1024 / 1024

    if mp4_size_mb > 0:
        print(f"MP4 检查通过：{OUTPUT_MP4}")
        print(f"MP4 文件大小：{mp4_size_mb:.2f} MB")
    else:
        print("MP4 文件存在，但大小为 0，生成失败。")
else:
    print("没有检测到 MP4 文件。")


# =====================================================
# 12. 检查 GIF 是否是真动图
# =====================================================

if GENERATE_GIF and os.path.exists(OUTPUT_GIF):
    try:
        img = Image.open(OUTPUT_GIF)

        frame_count = getattr(img, "n_frames", 1)
        file_format = img.format
        file_size_mb = os.path.getsize(OUTPUT_GIF) / 1024 / 1024

        print("\nGIF 文件格式：", file_format)
        print("GIF 帧数：", frame_count)
        print(f"GIF 文件大小：{file_size_mb:.2f} MB")

        if file_format == "GIF" and frame_count > 1:
            print("GIF 检查通过：这是可以播放的 GIF 动图。")
        else:
            print("GIF 检查失败：这不是有效 GIF 动图。")

    except Exception as e:
        print("GIF 检查失败。")
        print("原因：", e)


# =====================================================
# 13. 输出数据摘要
# =====================================================

print("\n温度汇总数据：")
print(summary.round(2).to_string(index=False))


# =====================================================
# 14. 使用建议
# =====================================================

print("\n使用建议：")
print("1. 发给学姐时，优先发送 animated_glass_process_curve.mp4。")
print("2. PPT 中建议插入 MP4，播放更稳定。")
print("3. 如果必须使用 GIF，把 GENERATE_GIF 改成 True 后重新运行。")
print("4. 不要把静态截图改后缀成 .gif，那样不会动。")
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager
from matplotlib.patches import Rectangle


# =====================================================
# 1. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    selected_font = None
    for font in candidate_fonts:
        if font in available_fonts:
            selected_font = font
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False

    if selected_font:
        print(f"当前使用字体：{selected_font}")
    else:
        print("未找到常见中文字体，若中文乱码，请安装 Microsoft YaHei 或 SimHei。")


set_chinese_font()


# =====================================================
# 2. 文件路径
# =====================================================

CSV_PATH = "glass_experiment_numeric_only.csv"
OUTPUT_PATH = "low_temperature_process_window_dashboard.png"


# =====================================================
# 3. 读取数据
# =====================================================

if not os.path.exists(CSV_PATH):
    raise FileNotFoundError(
        f"没有找到数据文件：{CSV_PATH}\n"
        f"请把 glass_experiment_numeric_only.csv 放在当前脚本同一目录。"
    )

df = pd.read_csv(CSV_PATH)

required_cols = [
    "temperature_c",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overheat_score",
    "overall_quality_score_100"
]

missing_cols = [col for col in required_cols if col not in df.columns]
if missing_cols:
    raise ValueError("CSV 缺少必要字段：" + "、".join(missing_cols))

for col in required_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()


# =====================================================
# 4. 指标归一化与工艺指标构建
# =====================================================
# 说明：
# particle_score、volume_score、transparency_score、success_score、overheat_score 原始按 0-5 计
# overall_quality_score_100 原始按 0-100 计
#
# overheat_score 是负向指标，数值越高代表过热越明显
# 因此这里转换为 overheat_control，越高代表控过热能力越好
# =====================================================

df["颗粒感"] = df["particle_score"] / 5 * 100
df["体积感"] = df["volume_score"] / 5 * 100
df["透光度"] = df["transparency_score"] / 5 * 100
df["成功度"] = df["success_score"] / 5 * 100
df["控过热能力"] = 100 - df["overheat_score"] / 5 * 100
df["综合质量"] = df["overall_quality_score_100"]

# 工艺稳定性：四个正向质量指标的均值
df["工艺稳定性"] = df[["颗粒感", "体积感", "透光度", "成功度"]].mean(axis=1)

# 低温适配指数：综合质量 + 控过热能力 + 工艺稳定性
df["低温适配指数"] = (
    df["综合质量"] * 0.45
    + df["控过热能力"] * 0.25
    + df["工艺稳定性"] * 0.30
)

summary = (
    df.groupby("temperature_c")
    .agg(
        样本数=("temperature_c", "count"),
        综合质量=("综合质量", "mean"),
        颗粒感=("颗粒感", "mean"),
        体积感=("体积感", "mean"),
        透光度=("透光度", "mean"),
        成功度=("成功度", "mean"),
        控过热能力=("控过热能力", "mean"),
        工艺稳定性=("工艺稳定性", "mean"),
        低温适配指数=("低温适配指数", "mean")
    )
    .reset_index()
    .sort_values("temperature_c")
)

temps = summary["temperature_c"].values
best_row = summary.loc[summary["低温适配指数"].idxmax()]
best_temp = int(best_row["temperature_c"])


# =====================================================
# 5. 创建画布
# =====================================================

fig = plt.figure(figsize=(18, 10), dpi=300)
fig.patch.set_facecolor("white")

ax_title = fig.add_axes([0.05, 0.86, 0.90, 0.10])
ax_main = fig.add_axes([0.08, 0.25, 0.58, 0.52])
ax_score = fig.add_axes([0.70, 0.29, 0.23, 0.44])
ax_footer = fig.add_axes([0.05, 0.05, 0.90, 0.13])

for ax in [ax_title, ax_footer]:
    ax.axis("off")


# =====================================================
# 6. 标题区
# =====================================================

ax_title.text(
    0.0,
    0.72,
    "低温技术：再生玻璃烧制工艺窗口优化图",
    fontsize=24,
    fontweight="bold",
    ha="left",
    va="center"
)

ax_title.text(
    0.0,
    0.23,
    "围绕温度变化同步评估综合质量、颗粒感、体积感、透光度、成功度与过热控制能力",
    fontsize=14,
    fontweight="bold",
    ha="left",
    va="center"
)

ax_title.text(
    1.0,
    0.28,
    f"推荐工艺窗口：{best_temp}℃ 附近",
    fontsize=13,
    fontweight="bold",
    ha="right",
    va="center",
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="black",
        linewidth=1.3
    )
)


# =====================================================
# 7. 主图：多指标工艺窗口图
# =====================================================
# 这个图不是普通折线图：
# - 背景区间表达温度风险
# - 多条指标线表达质量变化
# - 气泡大小表达样本数
# - 粗线表达低温适配指数
# =====================================================

# 背景风险区间
ax_main.axvspan(790, 830, alpha=0.08)
ax_main.axvspan(770, 790, alpha=0.05)
ax_main.axvspan(740, 770, alpha=0.08)

ax_main.text(
    810,
    106,
    "高温过熔风险区",
    ha="center",
    va="center",
    fontsize=10,
    fontweight="bold"
)

ax_main.text(
    780,
    106,
    "过渡调整区",
    ha="center",
    va="center",
    fontsize=10,
    fontweight="bold"
)

ax_main.text(
    755,
    106,
    "低温优化区",
    ha="center",
    va="center",
    fontsize=10,
    fontweight="bold"
)

metric_cols = ["综合质量", "颗粒感", "体积感", "透光度", "成功度", "控过热能力"]

for col in metric_cols:
    ax_main.plot(
        summary["temperature_c"],
        summary[col],
        marker="o",
        linewidth=2.0,
        markersize=5,
        label=col
    )

# 低温适配指数重点加粗
ax_main.plot(
    summary["temperature_c"],
    summary["低温适配指数"],
    marker="D",
    linewidth=3.6,
    markersize=7,
    linestyle="--",
    label="低温适配指数"
)

# 样本数气泡
bubble_size = summary["样本数"] * 90
ax_main.scatter(
    summary["temperature_c"],
    summary["低温适配指数"],
    s=bubble_size,
    facecolors="none",
    edgecolors="black",
    linewidths=1.4,
    label="样本规模"
)

# 最优点标注
ax_main.scatter(
    best_row["temperature_c"],
    best_row["低温适配指数"],
    s=260,
    marker="*",
    edgecolors="black",
    linewidths=1.2,
    zorder=5
)

ax_main.text(
    best_row["temperature_c"],
    best_row["低温适配指数"] + 7,
    f"推荐温度：{best_temp}℃",
    ha="center",
    va="bottom",
    fontsize=11,
    fontweight="bold",
    bbox=dict(
        boxstyle="round,pad=0.35",
        facecolor="white",
        edgecolor="black",
        linewidth=1.1
    )
)

ax_main.set_title(
    "温度-质量-稳定性多指标响应窗口",
    fontsize=16,
    fontweight="bold",
    pad=14
)

ax_main.set_xlabel("烧制温度（℃）", fontsize=12, fontweight="bold")
ax_main.set_ylabel("指标得分（0-100）", fontsize=12, fontweight="bold")
ax_main.set_ylim(0, 112)
ax_main.set_xlim(summary["temperature_c"].min() - 8, summary["temperature_c"].max() + 8)
ax_main.set_xticks(summary["temperature_c"].values)
ax_main.grid(True, linestyle="--", alpha=0.35)

ax_main.legend(
    loc="lower left",
    fontsize=8.8,
    frameon=True,
    ncol=2
)


# =====================================================
# 8. 右侧：低温技术综合评分条形图
# =====================================================

score_items = {
    "原料适配": 85,
    "节能降耗": 88,
    "过热控制": best_row["控过热能力"],
    "成品稳定": best_row["工艺稳定性"],
    "艺术肌理": best_row["颗粒感"],
    "产品转化": best_row["综合质量"],
}

score_names = list(score_items.keys())
score_values = np.array(list(score_items.values()), dtype=float)

y_pos = np.arange(len(score_names))

ax_score.barh(y_pos, score_values)

ax_score.set_yticks(y_pos)
ax_score.set_yticklabels(score_names, fontsize=10, fontweight="bold")
ax_score.set_xlim(0, 100)
ax_score.set_xlabel("评分（0-100）", fontsize=11, fontweight="bold")
ax_score.set_title("低温技术价值评分", fontsize=15, fontweight="bold", pad=12)
ax_score.grid(axis="x", linestyle="--", alpha=0.35)

for y, value in zip(y_pos, score_values):
    ax_score.text(
        value + 1.2,
        y,
        f"{value:.1f}",
        va="center",
        fontsize=9.5,
        fontweight="bold"
    )


# =====================================================
# 9. 底部说明
# =====================================================

summary_text = (
    f"分析结论：从现有实验数据看，{best_temp}℃附近的低温烧制条件在综合质量、过热控制和工艺稳定性上表现更优。"
    "相比高温烧制，低温技术能够减少过熔风险，保留玻璃颗粒边界、体积感和透光层次，"
    "更适合再生玻璃文创产品的小批量试制与艺术化转化。"
)

ax_footer.text(
    0.0,
    0.72,
    summary_text,
    ha="left",
    va="center",
    fontsize=11.2,
    bbox=dict(
        boxstyle="round,pad=0.45",
        facecolor="white",
        edgecolor="black",
        linewidth=1.1
    )
)

ax_footer.text(
    0.0,
    0.20,
    "数据来源：glass_experiment_numeric_only.csv；指标说明：低温适配指数 = 综合质量×0.45 + 控过热能力×0.25 + 工艺稳定性×0.30。",
    ha="left",
    va="center",
    fontsize=9.6
)


# =====================================================
# 10. 保存
# =====================================================

plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight", facecolor="white")
plt.show()

print(f"已保存：{OUTPUT_PATH}")
print("\n温度汇总数据：")
print(summary.round(2).to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Wedge, Circle, FancyBboxPatch
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
from matplotlib import font_manager


# =========================
# 1. 文件路径
# =========================
CSV_PATH = "glass_experiment_numeric_only.csv"
OUTPUT_PATH = "low_temp_circular_metric_map.png"


# =========================
# 2. 中文字体
# =========================
def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]
    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    for font in candidate_fonts:
        if font in available_fonts:
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False


set_chinese_font()


# =========================
# 3. 读取与处理数据
# =========================
df = pd.read_csv(CSV_PATH)

numeric_cols = [
    "temperature_c",
    "particle_score",
    "volume_score",
    "transparency_score",
    "success_score",
    "overheat_score",
    "overall_quality_score_100"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["temperature_c", "overall_quality_score_100"]).copy()

# 统一归一化到 0-100
df["particle_norm"] = df["particle_score"] / 5 * 100
df["volume_norm"] = df["volume_score"] / 5 * 100
df["transparency_norm"] = df["transparency_score"] / 5 * 100
df["success_norm"] = df["success_score"] / 5 * 100
df["overheat_control_norm"] = 100 - df["overheat_score"] / 5 * 100
df["overall_norm"] = df["overall_quality_score_100"]

summary = (
    df.groupby("temperature_c")
    .agg(
        overall=("overall_norm", "mean"),
        particle=("particle_norm", "mean"),
        volume=("volume_norm", "mean"),
        transparency=("transparency_norm", "mean"),
        success=("success_norm", "mean"),
        overheat_control=("overheat_control_norm", "mean")
    )
    .reset_index()
)

# 固定温度顺序：内圈 760，中圈 780，外圈 800
temperature_order = [760, 780, 800]
summary = (
    summary.set_index("temperature_c")
    .reindex(temperature_order)
    .dropna()
    .reset_index()
)

temps = summary["temperature_c"].astype(int).tolist()

metrics = [
    "overall",
    "particle",
    "volume",
    "transparency",
    "success",
    "overheat_control"
]

metric_labels = [
    "综合质量\nOverall",
    "颗粒感\nParticle",
    "体积感\nVolume",
    "透光度\nTransparency",
    "成功度\nSuccess",
    "控过热\nOverheat Control"
]

data = summary[metrics].values  # shape = (3, 6)


# =========================
# 4. 创建画布
# =========================
fig, ax = plt.subplots(figsize=(12, 12), dpi=300)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")
ax.set_aspect("equal")
ax.axis("off")

ax.set_xlim(-1.55, 1.70)
ax.set_ylim(-1.55, 1.55)

# 配色：专业、清晰
cmap = plt.cm.YlGnBu
norm = Normalize(vmin=0, vmax=100)

# 环形参数
n_rings = len(temps)
n_sectors = len(metrics)

inner_radius = 0.28
ring_width = 0.22
ring_gap = 0.02
sector_gap_deg = 3
start_angle = 90
sector_angle = 360 / n_sectors

# =========================
# 5. 标题
# =========================
ax.text(
    0,
    1.40,
    "低温烧制多指标环形热力图",
    ha="center",
    va="center",
    fontsize=24,
    fontweight="bold",
    color="black"
)

ax.text(
    0,
    1.27,
    "Low-Temperature Fusing Circular Metric Map",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold",
    color="black"
)


# =========================
# 6. 绘制环形热力图
# =========================
for i, temp in enumerate(temps):
    r0 = inner_radius + i * (ring_width + ring_gap)
    r1 = r0 + ring_width

    for j, metric in enumerate(metrics):
        theta2 = start_angle - j * sector_angle - sector_gap_deg / 2
        theta1 = start_angle - (j + 1) * sector_angle + sector_gap_deg / 2

        value = float(data[i, j])

        wedge = Wedge(
            center=(0, 0),
            r=r1,
            theta1=theta1,
            theta2=theta2,
            width=ring_width,
            facecolor=cmap(norm(value)),
            edgecolor="white",
            linewidth=2.0
        )
        ax.add_patch(wedge)

        # 数值标注
        theta_mid = np.deg2rad((theta1 + theta2) / 2)
        r_mid = (r0 + r1) / 2
        x = r_mid * np.cos(theta_mid)
        y = r_mid * np.sin(theta_mid)

        text_color = "white" if value >= 58 else "black"

        ax.text(
            x,
            y,
            f"{value:.0f}",
            ha="center",
            va="center",
            fontsize=10,
            fontweight="bold",
            color=text_color
        )


# =========================
# 7. 中心圆
# =========================
center_circle = Circle(
    (0, 0),
    inner_radius - 0.03,
    facecolor="#0B1F33",
    edgecolor="black",
    linewidth=1.5
)
ax.add_patch(center_circle)

ax.text(
    0,
    0.06,
    "低温技术",
    ha="center",
    va="center",
    fontsize=20,
    fontweight="bold",
    color="white"
)

ax.text(
    0,
    -0.06,
    "多指标评估",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold",
    color="white"
)

ax.text(
    0,
    -0.16,
    "Multi-Metric\nEvaluation",
    ha="center",
    va="center",
    fontsize=9.5,
    color="white"
)


# =========================
# 8. 外圈指标标签
# =========================
outer_label_radius = inner_radius + n_rings * (ring_width + ring_gap) + 0.18

for j, label in enumerate(metric_labels):
    theta = np.deg2rad(start_angle - (j + 0.5) * sector_angle)
    x = outer_label_radius * np.cos(theta)
    y = outer_label_radius * np.sin(theta)

    if x > 0.20:
        ha = "left"
    elif x < -0.20:
        ha = "right"
    else:
        ha = "center"

    ax.text(
        x,
        y,
        label,
        ha=ha,
        va="center",
        fontsize=11,
        fontweight="bold",
        color="black"
    )


# =========================
# 9. 温度环带说明
# =========================
box_x = -1.42
box_y = 1.02
box_w = 0.55
box_h = 0.35

info_box = FancyBboxPatch(
    (box_x, box_y),
    box_w,
    box_h,
    boxstyle="round,pad=0.02,rounding_size=0.02",
    linewidth=1.2,
    edgecolor="black",
    facecolor="white"
)
ax.add_patch(info_box)

ax.text(
    box_x + box_w / 2,
    box_y + box_h - 0.08,
    "环带对应温度",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold"
)

ax.text(
    box_x + 0.06,
    box_y + 0.17,
    "内圈：760℃\n中圈：780℃\n外圈：800℃",
    ha="left",
    va="top",
    fontsize=10,
    fontweight="bold",
    linespacing=1.6
)


# =========================
# 10. 颜色条
# =========================
sm = ScalarMappable(norm=norm, cmap=cmap)
sm.set_array([])

cbar = fig.colorbar(
    sm,
    ax=ax,
    fraction=0.045,
    pad=0.04
)

cbar.set_label(
    "标准化得分 / Normalized Score",
    fontsize=11,
    fontweight="bold"
)

cbar.ax.tick_params(labelsize=9)


# =========================
# 11. 输出
# =========================
plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch
from matplotlib import font_manager


# =====================================================
# 1. 中文字体设置
# =====================================================

def set_chinese_font():
    candidate_fonts = [
        "Microsoft YaHei",
        "SimHei",
        "Noto Sans CJK SC",
        "Source Han Sans SC",
        "Arial Unicode MS"
    ]

    available_fonts = set(f.name for f in font_manager.fontManager.ttflist)

    for font in candidate_fonts:
        if font in available_fonts:
            plt.rcParams["font.sans-serif"] = [font]
            break

    plt.rcParams["axes.unicode_minus"] = False


set_chinese_font()


# =====================================================
# 2. 输出路径
# =====================================================

OUTPUT_PATH = "solution_network_identity_bilingual_logic_fixed.png"


# =====================================================
# 3. 画布设置
# =====================================================

fig, ax = plt.subplots(figsize=(12, 10), dpi=300)
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

ax.set_xlim(-6.8, 6.8)
ax.set_ylim(-6.2, 6.2)
ax.axis("off")

black = "#000000"
gray = "#333333"


# =====================================================
# 4. 基础绘图函数
# =====================================================

def add_text(
    x,
    y,
    text,
    size=14,
    weight="bold",
    color=black,
    ha="center",
    va="center",
    rotation=0
):
    ax.text(
        x,
        y,
        text,
        fontsize=size,
        fontweight=weight,
        color=color,
        ha=ha,
        va=va,
        rotation=rotation
    )


def curved_arrow(
    start,
    end,
    rad=0.2,
    lw=1.35,
    mutation=14,
    text="",
    text_xy=None,
    text_size=8.2,
    rotation=0,
    alpha=1.0,
    style="->",
    linestyle="-"
):
    arrow = FancyArrowPatch(
        start,
        end,
        connectionstyle=f"arc3,rad={rad}",
        arrowstyle=style,
        mutation_scale=mutation,
        linewidth=lw,
        color=black,
        alpha=alpha,
        linestyle=linestyle
    )
    ax.add_patch(arrow)

    if text and text_xy:
        add_text(
            text_xy[0],
            text_xy[1],
            text,
            size=text_size,
            weight="normal",
            color=black,
            rotation=rotation
        )


# =====================================================
# 5. 核心节点坐标
# =====================================================

pos = {
    "market": (0, 5.45),
    "culture": (0, 3.75),
    "recycle": (-4.9, 0.75),
    "identity": (0, 0.25),
    "brand": (4.9, 0.75),
    "image": (0, -2.55),
    "time": (0, -5.05),

    "lowcarbon": (-5.45, 2.35),
    "aesthetic": (-3.25, 3.75),
    "waste_recovery": (-3.75, -1.35),
    "retail": (3.75, -1.35),
    "education": (5.25, 2.35),
    "industrial": (3.1, 3.15),
}


# =====================================================
# 6. 主节点文字
# =====================================================

add_text(
    *pos["market"],
    "用户与市场\nUsers & Market",
    size=17
)

add_text(
    *pos["culture"],
    "工艺文化\nCraft-Culture",
    size=15
)

add_text(
    *pos["recycle"],
    "废料回收\nWaste Glass\nRecycling",
    size=16
)

add_text(
    *pos["identity"],
    "绿色价值认同\nGreen Value\nIdentity",
    size=18
)

add_text(
    *pos["brand"],
    "品牌方案\nBrand Solution",
    size=17
)

add_text(
    *pos["image"],
    "产品形象\nProduct Image",
    size=16
)

add_text(
    pos["time"][0],
    pos["time"][1],
    "时间维度 / Temporal Dimensions:\n"
    "研发迭代（R&D） · 产业应用（Application） · 市场反馈（Feedback）",
    size=10.5,
    weight="normal"
)


# =====================================================
# 7. 中心说明
# =====================================================

add_text(
    0,
    2.65,
    "低温研发、美学设计、废料回收与市场反馈\n"
    "共同构成方案闭环与价值网络\n"
    "Low-carbon R&D, aesthetic design, recycling and feedback\n"
    "form a closed-loop solution network.",
    size=9.6,
    weight="normal",
    color=gray
)

add_text(
    0,
    -0.72,
    "以再生玻璃为媒介，连接生产端、消费端与品牌端，\n"
    "实现废料资源向绿色文创产品的价值跃升。\n"
    "Recycled glass connects production, consumption and branding,\n"
    "transforming waste resources into green cultural products.",
    size=9.4,
    weight="normal",
    color=gray
)


# =====================================================
# 8. 六个关键词节点文字
# =====================================================

add_text(
    *pos["lowcarbon"],
    "低碳研发\nLow-Carbon R&D",
    size=10.5,
    weight="bold"
)

add_text(
    *pos["aesthetic"],
    "美学设计\nAesthetic Design",
    size=10.5,
    weight="bold"
)

add_text(
    *pos["waste_recovery"],
    "废料回收\nWaste Recovery",
    size=10.5,
    weight="bold"
)

add_text(
    *pos["retail"],
    "零售转化\nRetail Conversion",
    size=10.5,
    weight="bold"
)

add_text(
    *pos["education"],
    "社会美育\nPublic Aesthetic Education",
    size=10.2,
    weight="bold"
)

add_text(
    *pos["industrial"],
    "产业应用\nIndustrial Application",
    size=10.5,
    weight="bold"
)


# =====================================================
# 9. 修正后的核心逻辑箭头
# =====================================================

# -----------------------------------------------------
# 用户与市场：需求源头与反馈终点
# -----------------------------------------------------

# 用户与市场 -> 工艺文化
curved_arrow(
    (0, 5.12),
    (0, 4.18),
    rad=0,
    lw=1.45,
    text="需求牵引\nDemand-driven",
    text_xy=(0.85, 4.65),
    text_size=7.8
)

# 用户与市场 -> 品牌方案
curved_arrow(
    (0.45, 5.15),
    (4.45, 1.45),
    rad=-0.24,
    lw=1.35,
    text="市场需求塑造品牌方向\nMarket demand shapes brand",
    text_xy=(2.7, 4.15),
    text_size=7.8,
    rotation=31
)

# 品牌方案 -> 用户与市场，反馈回路
curved_arrow(
    (5.15, 1.15),
    (0.75, 5.18),
    rad=0.36,
    lw=1.35,
    text="品牌反馈市场\nBrand feedback",
    text_xy=(4.1, 4.55),
    text_size=7.6,
    rotation=-33
)


# -----------------------------------------------------
# 低碳研发：上游能力
# -----------------------------------------------------

# 低碳研发 -> 废料回收
curved_arrow(
    (-5.25, 2.05),
    (-5.02, 1.22),
    rad=0.12,
    lw=1.2,
    text="提供低温工艺基础\nProvides process basis",
    text_xy=(-5.9, 1.72),
    text_size=7.4,
    rotation=76
)

# 低碳研发 -> 美学设计
curved_arrow(
    (-5.05, 2.65),
    (-3.75, 3.55),
    rad=0.08,
    lw=1.2,
    text="技术支撑设计\nSupports design",
    text_xy=(-4.55, 3.35),
    text_size=7.4,
    rotation=29
)

# 低碳研发 -> 产业应用
curved_arrow(
    (-4.95, 2.35),
    (2.45, 3.05),
    rad=-0.08,
    lw=1.15,
    text="导向应用验证\nLeads to application",
    text_xy=(-1.1, 3.25),
    text_size=7.4
)


# -----------------------------------------------------
# 设计、工艺、回收：构成价值基础
# -----------------------------------------------------

# 美学设计 -> 工艺文化
curved_arrow(
    (-2.75, 3.75),
    (-0.75, 3.75),
    rad=0,
    lw=1.25,
    text="表达再生美学\nExpressing aesthetics",
    text_xy=(-1.75, 4.08),
    text_size=7.4
)

# 废料回收 -> 工艺文化
curved_arrow(
    (-4.35, 1.08),
    (-0.75, 3.45),
    rad=-0.08,
    lw=1.25,
    text="提供材料与工艺来源\nMaterial and process source",
    text_xy=(-2.65, 2.35),
    text_size=7.4,
    rotation=30
)

# 废料回收 -> 绿色价值认同
curved_arrow(
    (-3.95, 0.65),
    (-1.05, 0.33),
    rad=0,
    lw=1.35,
    text="提供材料基础\nProvides material base",
    text_xy=(-2.45, 0.82),
    text_size=7.5
)

# 废料回收 Waste Recovery -> Waste Glass Recycling
curved_arrow(
    (-3.75, -1.05),
    (-4.55, 0.35),
    rad=-0.18,
    lw=1.15,
    text="回收强化\nReinforces recycling",
    text_xy=(-4.55, -0.65),
    text_size=7.3,
    rotation=72
)

# 废料回收 Waste Recovery -> 绿色价值认同
curved_arrow(
    (-3.35, -1.15),
    (-0.85, -0.05),
    rad=0.08,
    lw=1.15,
    text="强化再生价值理解\nReinforces value",
    text_xy=(-2.35, -0.82),
    text_size=7.3,
    rotation=18
)


# -----------------------------------------------------
# 工艺文化、产业应用：进入认同与品牌
# -----------------------------------------------------

# 工艺文化 -> 绿色价值认同
curved_arrow(
    (0, 3.28),
    (0, 0.95),
    rad=0,
    lw=1.35,
    text="塑造 / Shapes",
    text_xy=(0.65, 2.10),
    text_size=7.8
)

# 工艺文化 -> 品牌方案
curved_arrow(
    (0.75, 3.55),
    (4.35, 1.18),
    rad=-0.10,
    lw=1.25,
    text="形成品牌叙事\nBuilds narrative",
    text_xy=(2.75, 2.55),
    text_size=7.4,
    rotation=-31
)

# 工艺文化 -> 产业应用
curved_arrow(
    (0.75, 3.55),
    (2.55, 3.18),
    rad=-0.05,
    lw=1.15,
    text="转入应用场景\nTo application",
    text_xy=(1.68, 3.58),
    text_size=7.2
)

# 产业应用 -> 绿色价值认同
curved_arrow(
    (2.75, 2.85),
    (0.85, 0.72),
    rad=0.05,
    lw=1.15,
    text="应用验证价值\nValidates value",
    text_xy=(1.95, 1.65),
    text_size=7.3,
    rotation=-43
)

# 产业应用 -> 品牌方案
curved_arrow(
    (3.35, 2.88),
    (4.55, 1.35),
    rad=-0.08,
    lw=1.2,
    text="支撑品牌落地\nSupports landing",
    text_xy=(4.18, 2.15),
    text_size=7.3,
    rotation=-58
)


# -----------------------------------------------------
# 绿色价值认同：中心枢纽
# -----------------------------------------------------

# 绿色价值认同 -> 品牌方案
curved_arrow(
    (1.0, 0.25),
    (4.05, 0.62),
    rad=0,
    lw=1.35,
    text="建构价值表达\nConstructs value",
    text_xy=(2.55, 0.55),
    text_size=7.5
)

# 绿色价值认同 -> 产品形象
curved_arrow(
    (0, -0.35),
    (0, -2.05),
    rad=0,
    lw=1.35,
    text="映射 / Mirroring",
    text_xy=(-0.85, -1.18),
    text_size=7.8,
    rotation=90
)


# -----------------------------------------------------
# 产品形象、零售转化、品牌方案：商业闭环
# -----------------------------------------------------

# 产品形象 -> 零售转化
curved_arrow(
    (0.75, -2.55),
    (3.25, -1.55),
    rad=0.08,
    lw=1.25,
    text="形成消费触点\nCreates retail touchpoint",
    text_xy=(2.0, -2.20),
    text_size=7.3,
    rotation=19
)

# 零售转化 -> 品牌方案
curved_arrow(
    (3.85, -1.02),
    (4.75, 0.28),
    rad=0.05,
    lw=1.25,
    text="销售反馈品牌\nRetail feedback",
    text_xy=(4.65, -0.45),
    text_size=7.3,
    rotation=72
)


# -----------------------------------------------------
# 社会美育：扩散与反馈
# -----------------------------------------------------

# 社会美育 -> 品牌方案
curved_arrow(
    (5.10, 2.05),
    (5.05, 1.25),
    rad=-0.08,
    lw=1.2,
    text="增强公共认同\nEnhances recognition",
    text_xy=(5.82, 1.66),
    text_size=7.3,
    rotation=78
)

# 社会美育 -> 用户与市场
curved_arrow(
    (5.25, 2.65),
    (0.60, 5.42),
    rad=0.32,
    lw=1.20,
    text="美育扩散至市场\nEducation feeds market",
    text_xy=(3.45, 5.08),
    text_size=7.2,
    rotation=-20
)


# -----------------------------------------------------
# 时间维度：辅助背景，不作为主导节点
# -----------------------------------------------------

curved_arrow(
    (-4.65, -1.85),
    (-0.95, -4.65),
    rad=0.30,
    lw=1.0,
    mutation=11,
    text="研发迭代支撑\nR&D iteration",
    text_xy=(-3.35, -3.65),
    text_size=7.2,
    rotation=-34,
    linestyle="--",
    alpha=0.85
)

curved_arrow(
    (4.65, -1.85),
    (0.95, -4.65),
    rad=-0.30,
    lw=1.0,
    mutation=11,
    text="市场反馈修正\nMarket feedback",
    text_xy=(3.35, -3.65),
    text_size=7.2,
    rotation=34,
    linestyle="--",
    alpha=0.85
)


# =====================================================
# 10. 标题
# =====================================================

fig.suptitle(
    "废玻再生方案价值网络图\nValue Network of Waste Glass Regeneration Solution",
    fontsize=20,
    fontweight="bold",
    y=0.98
)


# =====================================================
# 11. 保存
# =====================================================

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.savefig(OUTPUT_PATH, dpi=300, bbox_inches="tight", facecolor="white")
plt.show()

print(f"已保存：{OUTPUT_PATH}")
```
