```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

fig, ax = plt.subplots(figsize=(12,8))

ax.set_xlim(0,10)
ax.set_ylim(0,10)
ax.axis('off')

steps = [
    "Experimental\nData",
    "Process\nDatabase",
    "Feature\nEngineering",
    "ML Model",
    "Quality\nPrediction",
    "Visualization",
    "Decision\nRecommendation"
]

y = 9

for i,step in enumerate(steps):

    box = FancyBboxPatch(
        (3.2,y-0.7),
        3.5,
        0.9,
        boxstyle="round,pad=0.2",
        linewidth=2
    )

    ax.add_patch(box)

    ax.text(
        5,
        y-0.25,
        step,
        ha='center',
        va='center',
        fontsize=11
    )

    if i < len(steps)-1:

        ax.arrow(
            5,
            y-0.9,
            0,
            -0.6,
            head_width=0.12,
            head_length=0.15,
            length_includes_head=True
        )

    y -= 1.3

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(10,10))

ax.axis('off')

# 中心
ax.text(
    0,
    0,
    "Data\nCenter",
    fontsize=18,
    ha='center',
    va='center',
    bbox=dict(
        boxstyle="circle,pad=1",
        linewidth=3
    )
)

nodes = {
    "User\nProfile":(0,2.5),
    "Product\nUpgrade":(2.5,0),
    "Business\nConversion":(0,-2.5),
    "Market\nAnalysis":(-2.5,0),
    "Feedback\nLoop":(1.8,-1.8),
    "User\nGrowth":(-1.8,1.8)
}

for name,(x,y) in nodes.items():

    ax.text(
        x,
        y,
        name,
        ha='center',
        va='center',
        bbox=dict(
            boxstyle="round,pad=0.6",
            linewidth=2
        )
    )

    ax.arrow(
        0,
        0,
        x*0.8,
        y*0.8,
        head_width=0.12,
        length_includes_head=True
    )

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = [1,2,3,4,5,6]

users = [100,150,220,300,380,500]
feedback = [20,35,50,70,95,130]
revenue = [5,7,10,14,18,25]

fig, ax1 = plt.subplots(figsize=(10,6))

ax1.plot(months, users, 'o-', label='Users')
ax1.plot(months, feedback, 'o-', label='Feedback')

ax1.set_ylabel("User Metrics")

ax2 = ax1.twinx()
ax2.plot(months, revenue, 's--', color='red', label='Revenue')
ax2.set_ylabel("Revenue (10k)")

plt.title("User Feedback → Product → Market Growth")

fig.legend(loc="upper left")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = np.array([2018,2019,2020,2021,2022,2023,2026,2027,2028])

# 市场规模（代理）
market = np.array([50,55,60,68,80,95,110,125,140])

# 项目收入（模型输出）
revenue = np.array([10,12,14,18,22,28,35,42,50])

# 成本（递减）
cost = np.array([8,8.5,9,9.5,10,10.5,11,11.5,12])

# 利润
profit = revenue - cost

fig, ax1 = plt.subplots(figsize=(10,6))

# 市场 + 收入
ax1.plot(years, market, marker='o', label='Market Size', linewidth=2)
ax1.plot(years, revenue, marker='s', label='Project Revenue', linewidth=2)

ax1.set_xlabel("Year")
ax1.set_ylabel("Market / Revenue")

# 第二轴：利润
ax2 = ax1.twinx()
ax2.plot(years, profit, marker='^', linestyle='--', color='red', label='Profit')
ax2.set_ylabel("Profit")

plt.title("Market → Revenue → Profit Forecast (Structural Model)")

fig.legend(loc="upper left")

plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = ['2026','2027','2028']

base_profit = np.array([10,12,15])
ip_profit = np.array([3,5,8])        # IP授权
eco_profit = np.array([4,6,9])       # 再生材料溢价
cost_saving = np.array([2,3,4])      # 成本下降收益

width = 0.6

plt.figure(figsize=(10,6))

plt.bar(years, base_profit, label='Base Profit')
plt.bar(years, ip_profit, bottom=base_profit, label='IP Licensing')
plt.bar(years, eco_profit, bottom=base_profit+ip_profit, label='Eco Premium')
plt.bar(years, cost_saving, bottom=base_profit+ip_profit+eco_profit, label='Cost Saving')

plt.title("Profit Structure Decomposition (Stacked Model)")
plt.ylabel("Profit (10k RMB)")
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================
# Font
# ==========================

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

# ==========================
# Year
# ==========================

years = ['2026', '2027', '2028']

# ==========================
# 利润来源构成
# （结合国家统计局文化产业增长逻辑）
# ==========================

# 文创产品销售
product_sales = np.array([6, 8, 10])

# IP授权收益
ip_licensing = np.array([2, 4, 7])

# 环保艺术溢价
eco_premium = np.array([3, 5, 8])

# 废料替代节约成本
cost_saving = np.array([4, 6, 9])

# ==========================
# 总利润
# ==========================

total_profit = (
    product_sales
    + ip_licensing
    + eco_premium
    + cost_saving
)

# ==========================
# Figure
# ==========================

fig, ax = plt.subplots(
    figsize=(12, 7),
    dpi=120
)

# ==========================
# Stacked Bar
# ==========================

bar1 = ax.bar(
    years,
    product_sales,
    label='文创产品销售收益',
    linewidth=1
)

bar2 = ax.bar(
    years,
    ip_licensing,
    bottom=product_sales,
    label='IP授权收益',
    linewidth=1
)

bar3 = ax.bar(
    years,
    eco_premium,
    bottom=product_sales + ip_licensing,
    label='环保艺术溢价收益',
    linewidth=1
)

bar4 = ax.bar(
    years,
    cost_saving,
    bottom=product_sales + ip_licensing + eco_premium,
    label='工业废料降本收益',
    linewidth=1
)

# ==========================
# 显示总利润
# ==========================

for i, value in enumerate(total_profit):

    ax.text(
        i,
        value + 0.4,
        f'{value:.0f}',
        ha='center',
        fontsize=12,
        fontweight='bold'
    )

# ==========================
# 增长率
# ==========================

growth = [
    0,
    (total_profit[1]-total_profit[0])/total_profit[0]*100,
    (total_profit[2]-total_profit[1])/total_profit[1]*100
]

for i in range(1, len(years)):

    ax.annotate(
        f'+{growth[i]:.1f}%',
        xy=(i, total_profit[i]),
        xytext=(i, total_profit[i] + 3),
        ha='center',
        fontsize=10,
        arrowprops=dict(
            arrowstyle='->',
            lw=1.5
        )
    )

# ==========================
# Title
# ==========================

ax.set_title(
    '青橙焕艺利润增长驱动因素分析\nProfit Growth Drivers of Qingcheng Glass Art',
    fontsize=16,
    fontweight='bold',
    pad=20
)

# ==========================
# Axis
# ==========================

ax.set_ylabel(
    '利润（万元）',
    fontsize=12
)

ax.set_xlabel(
    '年份',
    fontsize=12
)

# ==========================
# Grid
# ==========================

ax.grid(
    axis='y',
    linestyle='--',
    alpha=0.3
)

# ==========================
# Legend
# ==========================

ax.legend(
    loc='upper left',
    fontsize=10,
    frameon=True
)

# ==========================
# Data Source
# ==========================

plt.figtext(
    0.5,
    0.01,
    '数据参考：国家统计局2026年一季度文化及相关产业企业数据（营业收入同比增长6.4%，文化新业态增长11.2%）',
    ha='center',
    fontsize=9
)

plt.tight_layout()

plt.show()
```
