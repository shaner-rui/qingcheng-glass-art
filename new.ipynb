```python
import pandas as pd

df = pd.read_csv("glass_experiment_numeric_only.csv")

# 1. 统一字段名（去除首尾空格，转小写）
df.columns = df.columns.str.strip().str.lower()

# 2. 重命名（按规范），注意正确闭合花括号
df = df.rename(columns={
    "record_id": "experiment_id",
    "date_yyyymmdd": "date",
    "temperature_c": "temperature_c",
    "hold_time_min": "hold_time_min",
    "particle_score": "particle_score",
    "volume_score": "volume_score",
    "transparency_score": "transparency_score",
    "overall_quality_score_100": "overall_quality_score"
})  # ← 这里必须有关闭的 } 和括号

# 3. 数值类型统一（单独处理）
numeric_cols = [
    "temperature_c",
    "hold_time_min",
    "particle_score",
    "volume_score",
    "transparency_score",
    "overall_quality_score"
]

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors="coerce")

# 4. 添加衍生指标
df["stability_score"] = (100 - abs(df["temperature_c"] - 760) * 0.2).clip(0, 100)
df["defect_rate"] = ((100 - df["overall_quality_score"]) / 100).clip(0, 1)

# 5. 保存结果
df.to_csv("glass_experiment_standardized.csv", index=False)

# 6. 查看前几行
print(df.head())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

fig, ax = plt.subplots(figsize=(7,7))
ax.set_xlim(-1.5,1.5)
ax.set_ylim(-1.5,1.5)
ax.axis("off")

# =========================
# 节点定义（循环系统）
# =========================
nodes = {
    "Waste Glass\nCollection": (0, 1.2),
    "Sorting\nProcessing": (1.2, 0),
    "Low-Temp\nMelting": (0, -1.2),
    "Artistic\nProducts": (-1.2, 0)
}

# =========================
# 画节点
# =========================
for name, (x, y) in nodes.items():
    ax.text(
        x, y, name,
        ha="center", va="center",
        fontsize=10,
        bbox=dict(
            boxstyle="round,pad=0.4",
            facecolor="#e8f8f5",
            edgecolor="#1abc9c",
            linewidth=2
        )
    )

# =========================
# 平滑循环箭头函数
# =========================
def curved_arrow(start, end, rad):
    arrow = FancyArrowPatch(
        start, end,
        connectionstyle=f"arc3,rad={rad}",
        arrowstyle='-|>',
        mutation_scale=15,
        lw=2,
        color="#27ae60"
    )
    ax.add_patch(arrow)

# =========================
# 循环路径（关键：圆滑）
# =========================
curved_arrow((0, 1.0), (1.0, 0.2), 0.3)
curved_arrow((1.0, -0.2), (0.2, -1.0), 0.3)
curved_arrow((-0.2, -1.0), (-1.0, -0.2), 0.3)
curved_arrow((-1.0, 0.2), (-0.2, 1.0), 0.3)

# =========================
# 中心系统标签
# =========================
ax.text(
    0, 0,
    "Closed-Loop\nGreen Glass System",
    ha="center", va="center",
    fontsize=12,
    fontweight="bold",
    bbox=dict(
        boxstyle="circle,pad=0.5",
        facecolor="#d5f5e3",
        edgecolor="#2ecc71",
        linewidth=2
    )
)

# =========================
# 能量/物质流标注
# =========================
ax.text(0, 1.55, "Material Flow + Energy Recycling", ha="center", fontsize=10, color="#239b56")

plt.title("Green Circular Technology System Architecture", fontsize=14)
plt.show()
```
